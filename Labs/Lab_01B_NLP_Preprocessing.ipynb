```python
import nltk
import pandas as pd
from nltk.tokenize import word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer

print('Libraries imported successfully.')
```

```python
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
print('NLTK resources are ready.')
```

## 1. Given Text Corpus

We use a small corpus containing repeated words, different cases, punctuation, and common stop words.

```python
corpus = [
    'Natural Language Processing helps computers understand human language.',
    'NLP techniques are used to analyze text, classify documents, and extract useful information.',
    'Students are learning NLP because language data is everywhere.',
    'Learning and understanding language helps build better NLP applications.'
]

for i, text in enumerate(corpus, start=1):
    print(f'Document {i}: {text}')
```

## 2. Tokenization

Tokenization breaks each sentence into individual words (tokens).

```python
tokens = [word_tokenize(text) for text in corpus]
print('Tokens from the first document:')
print(tokens[0])
```

## 3. Case Normalization

All words are converted to lowercase so that words such as `NLP` and `nlp` are treated consistently.

```python
normalized_tokens = [[word.lower() for word in doc] for doc in tokens]
print('Before normalization:')
print(tokens[0])
print('\nAfter normalization:')
print(normalized_tokens[0])
```

## 4. Stop-word Removal

Common words such as **the, is, are, and, to** are removed while meaningful words are retained.

```python
stop_words = set(stopwords.words('english'))
filtered_tokens = [[word for word in doc if word.isalpha() and word not in stop_words] for doc in normalized_tokens]

print('Before stop-word removal:')
print(normalized_tokens[0])
print('\nAfter stop-word removal:')
print(filtered_tokens[0])
```

## 5. Stemming

Stemming reduces words to a root-like form by changing or removing word endings.

```python
stemmer = PorterStemmer()
stemmed_tokens = [[stemmer.stem(word) for word in doc] for doc in filtered_tokens]

print('Before stemming:')
print(filtered_tokens[0])
print('\nAfter stemming:')
print(stemmed_tokens[0])
```

## 6. Lemmatization

Lemmatization converts words to their meaningful base or dictionary form.

```python
lemmatizer = WordNetLemmatizer()
lemmatized_tokens = [[lemmatizer.lemmatize(word) for word in doc] for doc in filtered_tokens]

print('Before lemmatization:')
print(filtered_tokens[0])
print('\nAfter lemmatization:')
print(lemmatized_tokens[0])
```

## 7. Compare the Preprocessing Steps

The first document is shown at each stage.

```python
i = 0
comparison = pd.DataFrame({
    'Stage': ['Original Text', 'Tokenization', 'Case Normalization', 'Stop-word Removal', 'Stemming', 'Lemmatization'],
    'Representation': [
        corpus[i],
        ' '.join(tokens[i]),
        ' '.join(normalized_tokens[i]),
        ' '.join(filtered_tokens[i]),
        ' '.join(stemmed_tokens[i]),
        ' '.join(lemmatized_tokens[i])
    ]
})
comparison
```

## 8. Effect on Text Representation

Compare the total number of tokens at different stages.

```python
counts = pd.DataFrame({
    'Stage': ['Tokenization', 'Case Normalization', 'Stop-word Removal', 'Stemming', 'Lemmatization'],
    'Total Tokens': [
        sum(len(d) for d in tokens),
        sum(len(d) for d in normalized_tokens),
        sum(len(d) for d in filtered_tokens),
        sum(len(d) for d in stemmed_tokens),
        sum(len(d) for d in lemmatized_tokens)
    ]
})
counts
```

### Observation

- Tokenization breaks text into individual units.
- Case normalization makes representation consistent.
- Stop-word removal reduces less-informative words.
- Stemming produces root-like forms.
- Lemmatization produces meaningful base forms.
- These steps can reduce noise and vocabulary variation, helping downstream NLP learning tasks.

## 9. Stemming vs Lemmatization

Compare both techniques on selected words.

```python
sample_words = ['playing', 'played', 'plays', 'studies', 'studying', 'better', 'running', 'applications']
comparison_words = pd.DataFrame({
    'Word': sample_words,
    'Stemmed': [stemmer.stem(w) for w in sample_words],
    'Lemmatized': [lemmatizer.lemmatize(w) for w in sample_words]
})
comparison_words
```
