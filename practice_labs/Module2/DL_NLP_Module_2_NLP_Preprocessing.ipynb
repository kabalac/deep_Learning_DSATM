## 3. Sample Text

We will use the same sentence through most examples.

````python
text = "The students are studying Deep Learning and NLP at DSATM!"
print(text)
````

## 4. Tokenization
**Tokenization** breaks text into smaller units called tokens.

````python
tokens = word_tokenize(text)
print(tokens)
````

### Simple Tokenization Example

````python
simple_text = "I love Deep Learning"
print(word_tokenize(simple_text))
````

## 5. Lowercasing

````python
lower_text = text.lower()
print(lower_text)
````

## 6. Remove Punctuation

````python
no_punctuation = lower_text.translate(
    str.maketrans('', '', string.punctuation)
)

print(no_punctuation)
````

### Remove punctuation after tokenization

````python
tokens = word_tokenize(no_punctuation)
tokens_without_punctuation = [t for t in tokens if t not in string.punctuation]
print(tokens_without_punctuation)
````

## 7. Stopword Removal

Stopwords are common words that are **sometimes** removed. The decision depends on the task.

````python
stop_words = set(stopwords.words("english"))
tokens = word_tokenize(text.lower())

filtered_tokens = [t for t in tokens if t not in stop_words]

print("Before:", tokens)
print("After :", filtered_tokens)
````

### Important: Stopwords are task-dependent

Consider: **"This movie is not good."**

Blindly removing `not` can change the meaning.

````python
example = "This movie is not good."
example_tokens = word_tokenize(example.lower())

print(example_tokens)
print("Stopwords:", [w for w in example_tokens if w in stop_words])
````

## 8. Stemming

````python
stemmer = PorterStemmer()

words = ["play", "playing", "played", "plays", "studies", "studying"]

for word in words:
    print(f"{word:10} -> {stemmer.stem(word)}")
````

## 9. Lemmatization

````python
lemmatizer = WordNetLemmatizer()

words = ["playing", "played", "plays", "studies", "studying"]

for word in words:
    print(f"{word:10} -> {lemmatizer.lemmatize(word, pos='v')}")
````

### Stemming vs Lemmatization

| Technique | Idea | Example |
|---|---|---|
| Stemming | Reduce toward a stem | `studies → studi` |
| Lemmatization | Find a meaningful base form | `studies → study` |

## 10. n-grams
An **n-gram** is a sequence of `n` consecutive tokens.

````python
sentence = "I love Natural Language Processing"
tokens = word_tokenize(sentence.lower())
print(tokens)
````

### Unigrams

````python
unigrams = [(t,) for t in tokens]
print(unigrams)
````

### Bigrams

````python
bigrams = list(zip(tokens, tokens[1:]))
print(bigrams)
````

### Trigrams

````python
trigrams = list(zip(tokens, tokens[1:], tokens[2:]))
print(trigrams)
````

## 11. Complete Preprocessing Pipeline

````python
def preprocess_text(text):
    # 1. Lowercase
    text = text.lower()

    # 2. Tokenize
    tokens = word_tokenize(text)

    # 3. Remove punctuation
    tokens = [t for t in tokens if t not in string.punctuation]

    # 4. Remove stopwords
    tokens = [t for t in tokens if t not in stop_words]

    # 5. Lemmatize
    tokens = [lemmatizer.lemmatize(t, pos="v") for t in tokens]

    return tokens

raw_text = "The students are studying Deep Learning and NLP at DSATM!"

print("RAW TEXT:")
print(raw_text)
print("\nPROCESSED TOKENS:")
print(preprocess_text(raw_text))
print("\nPROCESSED TEXT:")
print(" ".join(preprocess_text(raw_text)))
````

## 12. Processing a Small Corpus

````python
corpus = [
    "I love Deep Learning.",
    "Deep Learning is interesting.",
    "NLP is interesting and useful.",
    "The students are learning NLP."
]

for i, document in enumerate(corpus, 1):
    print(f"Document {i}: {document}")
````

````python
for document in corpus:
    print(preprocess_text(document))
````
